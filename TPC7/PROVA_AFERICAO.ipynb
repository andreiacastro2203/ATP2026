##### TPC 1: Especifique as seguintes listas em compreensão

a) Lista formada pelos elementos que não são comuns às duas listas:

````python
# Resultado esperado: [1,2,3,7,8]

lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]

ncomuns = []

# elementos de lista1 que não estão em lista2
for x in lista1:
    encontrado = False
    for y in lista2:
        if x == y:
            encontrado = True
    if encontrado == False:
        ncomuns.append(x)

# elementos de lista2 que não estão em lista1
for y in lista2:
    encontrado = False
    for x in lista1:
        if y == x:
            encontrado = True
    if encontrado == False:
        ncomuns.append(y)

print(ncomuns)   # [1, 2, 3, 6, 7, 8]
````

##### TPC 2: À semelhança do que foi feito nas aulas, realize as seguintes tarefas

a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

````python
def strCount(s, subs):
    contador = 0
    i = 0
    while i <= len(s) - len(subs):
        if s[i:i + len(subs)] == subs:
            contador = contador + 1
            i = i + len(subs)
        else:
            i = i + 1
    return contador

print(strCount("catcowcat", "cat"))   # 2
print(strCount("catcowcat", "cow"))   # 1
print(strCount("catcowcat", "dog"))   # 0
````

b) Especifique uma função que recebe uma lista de números inteiros positivos e devolve o menor produto que for possível calcular multiplicando os 3 menores inteiros da lista:

````python
def produtoM3(lista):
    ordenada = sorted(lista)
    produto = 1
    for i in range(3):
        produto = produto * ordenada[i]
    return produto

print(produtoM3([12, 3, 7, 10, 12, 8, 9]))   # 168
````

c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
def reduxInt(n):
    while n > 9:
        soma = 0
        while n > 0:
            soma = soma + (n % 10)
            n = n // 10
        n = soma
    return n

print(reduxInt(38))    # 2
print(reduxInt(777))   # 3
````

d) Especifique uma função que recebe duas strings, `string1` e `string2`, e devolve o índice da primeira ocorrência de `string2` em `string1`, caso não ocorra nenhuma vez a função deverá retornar `-1`:

````python
def myIndexOf(s1, s2):
    resultado = -1
    i = 0
    while resultado == -1 and i <= len(s1) - len(s2):
        if s1[i:i + len(s2)] == s2:
            resultado = i
        i = i + 1
    return resultado

print(myIndexOf("Hoje está um belo dia de sol!", "belo"))    # 13
print(myIndexOf("Hoje está um belo dia de sol!", "chuva"))   # -1
````

##### TPC 3: A Rede Social

À semelhança do que foi feito nas aulas, realize as seguintes tarefas

Considere que a informação sobre uma rede social está armazenada numa lista de dicionários.

Cada dicionário, correspondente a um _post_ e tem chaves `id`, `conteudo`, `autor`, `dataCriacao` e `comentarios`.
Por sua vez, `comentarios` é uma lista de dicionários com chaves `comentario` e `autor`.

Considere o seguinte exemplo:

``` 
    MyFaceBook = [{
        'id': 'p1', 
        'conteudo': 'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor
    tem de realizar...', 
        'autor': 'jcr', 
        'dataCriacao': '2023-07-20', 
        'comentarios': [
            {
                'comentario': 'Completamente de acordo...',
                'autor': 'prh'
            },
            {
                'comentario': 'Mas há quem goste...',
                'autor': 'jj'
            }
        ]},
        {
            'id': 'p2',
            ...
        },
        ...
        ]
```

Defina as seguintes funções de manipulação e consulta da rede social:

````python
MyFaceBook = [
    {'id': 'p1', 'conteudo': 'A tarefa de avaliação é talvez a mais ingrata...',
     'autor': 'jcr', 'dataCriacao': '2023-07-20',
     'comentarios': [{'comentario': 'Completamente de acordo...', 'autor': 'prh'},
                     {'comentario': 'Mas há quem goste...', 'autor': 'jj'}]},
    {'id': 'p2', 'conteudo': 'Hoje o dia está lindo.',
     'autor': 'prh', 'dataCriacao': '2023-07-21', 'comentarios': []},
    {'id': 'p3', 'conteudo': 'Mais um post meu.',
     'autor': 'jcr', 'dataCriacao': '2023-07-22',
     'comentarios': [{'comentario': 'Concordo', 'autor': 'prh'}]}
]
````

a) `quantosPost`, que indica quantos posts estão registados:

````python
def quantosPost(rede):
    return len(rede)

print("a) quantosPost:")
print(quantosPost(MyFaceBook))
````

b)  `postsAutor`, que devolve a lista de posts de um determinado autor:

````python
def postsAutor(rede, autor):
    res = []
    for post in rede:
        if post['autor'] == autor:
            res.append(post)
    return res

print("b) postsAutor('jcr'):")
print(postsAutor(MyFaceBook, 'jcr'))
````

c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

````python
def autores(redeSocial):
    lista = []
    for post in redeSocial:
        existe = False
        for a in lista:
            if a == post['autor']:
                existe = True
        if existe == False:
            lista.append(post['autor'])
    lista.sort()
    return lista

print("c) autores:")
print(autores(MyFaceBook))
````

d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 
    
O campo `id` devrá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(rede, conteudo, autor, dataCriacao, comentarios):
    maior = 0
    for post in rede:
        num = int(post['id'][1:])        # 'p12' -> 12
        if num > maior:
            maior = num
    novo = {
        'id': 'p' + str(maior + 1),
        'conteudo': conteudo,
        'autor': autor,
        'dataCriacao': dataCriacao,
        'comentarios': comentarios
        }
    rede.append(novo)
    return rede

print("d) insPost:")
MyFaceBook = insPost(MyFaceBook, 'Novo post', 'jj', '2023-08-01', [])
print(MyFaceBook)
````

e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(rede, id):
    nova = []
    for post in rede:
        if post['id'] != id:
            nova.append(post)
    return nova

print("e) remPost('p2'):")
MyFaceBook = remPost(MyFaceBook, 'p2')
print(MyFaceBook)
````

f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(rede):
    dist = {}
    for post in rede:
        a = post['autor']
        if a in dist:
            dist[a] += 1
        else:
            dist[a] = 1
    return dist

print("f) postsPorAutor:")
print(postsPorAutor(MyFaceBook))
````

g) `comentadoPor`, que recebe um autor e devolve a lista de posts comentados por esse autor.

````python
def comentadoPor(rede, autor):
    res = []
    for post in rede:
        encontrado = False
        for c in post['comentarios']:
            if c['autor'] == autor:
                encontrado = True
        if encontrado:
            res.append(post)
    return res

print("\ng) comentadoPor('prh'):")
print(comentadoPor(MyFaceBook, 'prh'))
````
